```python
import pandas as pd
from sklearn import metrics
from sklearn.model_selection import train_test_split
from sklearn.metrics import recall_score, classification_report, confusion_matrix
from sklearn.tree import DecisionTreeClassifier
from imblearn.combine import SMOTEENN
```

```python
df = pd.read_csv('tel_churn.csv')
```

```python
df.head()
```

```python
df = df.drop('Unnamed: 0',axis=1)
```

```python
df.head()
```

```python
# creating X and Y variables
x = df.drop('Churn' , axis=1)
x
```

```python
y = df['Churn']
y
```

```python
x_train , x_test , y_train,y_test = train_test_split(x,y,test_size=0.2)
```

### Decision Tree Classifier

```python
model_dt = DecisionTreeClassifier(criterion='gini' , random_state=100,max_depth=6,min_samples_split=8)
```

```python
model_dt.fit(x_train,y_train)
```

```python
y_pred = model_dt.predict(x_test)
```

```python
y_pred
```

```python
print(classification_report(y_test , y_pred , labels = [0,1]))
```

```python
print(confusion_matrix(y_test , y_pred))
```

```python
sm = SMOTEENN()
x_resampled, y_resampled = sm.fit_resample(x, y)
```

```python
xr_train , xr_test , yr_train,yr_test = train_test_split(x_resampled,y_resampled,test_size=0.2)
```

```python
model_dt_smote = DecisionTreeClassifier(criterion='gini' , random_state=100,max_depth=6,min_samples_split=8)
```

```python
model_dt_smote.fit(xr_train,yr_train)
```

```python
y_pred_smote = model_dt_smote.predict(xr_test)
```

```python
print(classification_report(yr_test , y_pred_smote , labels = [0,1]))
```

```python
print(confusion_matrix(yr_test , y_pred_smote))
```

```python
import pickle
```

```python
pickle.dump(model_dt_smote , open(filename , 'wb'))
```

```python
load_model = pickle.load(open(filename , 'rb'))
```

```python
load_model.score(xr_test ,yr_test)
```
